````python
coordinates = (10, 20)
# 튜플 언패킹
x,y = coordinates
print(f'x좌표: {x}, y좌표: {y}')

rgb = [255, 128, 0]
# 리스트 언패킹
red, green, blue = rgb
print(f'빨강: {red}, 초록: {green}, 파랑: {blue}')

word = 'ABC'
# 문자열 언패킹
first, second, third = word
print(f'첫 번째 문자: {first}, 두 번째 문자: {second}, 세 번째 문자: {third}')
````

````python
## 사용자 정의 함수 : 최소, 최대, 평균값 구하는 함수
def get_dataset_stats(values):
  return min(values), max(values), sum(values) / len(values)

data = [12, 8, 21, 17, 5, 32, 14]

# 함수값 return 언패킹
minimum, maximum, average = get_dataset_stats(data)
print(f'최솟값: {minimum}, 최댓값: {maximum}, 평균: {average:.2f}')
````

1. 함수 정의 (def get_dataset_stats(values):): 리스트를 입력받아 최솟값, 최댓값, 평균을 계산해 돌려주는 사용자 정의 함수를 먼저 만듭니다.

2. 데이터 선언 (data = [...]): 함수에 집어넣을 실제 숫자 리스트 데이터를 정의합니다.

3. 함수 실행 및 언패킹 (minimum, maximum, average = get_dataset_stats(data)): 만든 함수에 data를 전달해 실행하고, 한꺼번에 튀어나오는 3개의 결과값을 각각의 변수(minimum, maximum, average)에 딱 맞게 나누어 담습니다.

````python
# try 문
number = int(input('숫자를 입력하세요: '))

try:
    result = 10 / number
    print(f'결과: {result}')

# except 문 1
except ValueError:
    print('유효한 숫자를 입력해야 합니다.')

# except 문 2
except ZeroDivisionError:
    print('0으로 나눌 수 없습니다.')
````

여기서 number = int(input(...)) 코드가 try 블록 바깥(위쪽)에 나와 있으면, 사용자가 입력하는 단계에서부터 에러가 터지기 때문에 예외 처리의 보호를 받지 못하게 됩니다.

이유를 구체적으로 풀어보면 다음과 같습니다.

에러가 발생하는 위치: 사용자가 입력창에 숫자가 아니라 글자('ㅇ' 등)나 공백을 입력하면, int()로 바꾸는 과정에서 즉시 ValueError가 발생합니다.

try문의 역할: try문은 그 내부(try: 바로 아래 들여쓰기된 구문)에서 실행되는 코드에서 에러가 날 때만 잡아채서 except로 넘겨주는 역할을 합니다.

왜 밖으로 빼면 안 되는지: number를 입력받는 코드가 try 바깥에 있으면, try 안쪽으로 들어가기도 전에 int() 변환 과정에서 프로그램이 곧바로 멈춰버립니다. 결국 아래에 있는 except ValueError:가 "어? 에러 났네? 내가 잡아줄게!" 하고 손을 내밀기도 전에 파이썬 자체 경고창(Traceback)을 띄우며 죽어버리는 것이죠.

따라서 입력받는 과정(input과 int 변환) 자체도 try 블록 안으로 집어넣어야 사용자가 문자를 입력했을 때 ValueError를 정상적으로 잡아내서 "유효한 숫자를 입력해야 합니다."라는 안내 문구를 띄울 수 있습니다.

````python
# try 문
try:
    number = int(input('숫자를 입력하세요: '))
    result = 10 / number
    print(f'결과: {result}')

# except 문 1
except ValueError:
    print('유효한 숫자를 입력해야 합니다.')

# except 문 2
except ZeroDivisionError:
    print('0으로 나눌 수 없습니다.')
````

````python
# try 문
try:
    number = int(input('숫자를 입력하세요: '))
    result = 10 / number

# except 문 1
except ValueError:
    print('유효한 숫자를 입력해야 합니다.')

# except 문 2
except ZeroDivisionError:
    print('0으로 나눌 수 없습니다.')

print(f'결과: {result}')
````

사용자가 'ㅎ'을 입력하는 순간 int()에서 에러가 터져서 except ValueError:로 바로 점프합니다.

그래서 "유효한 숫자를 입력해야 합니다."가 먼저 출력됩니다.

그 후 try-except 블록이 끝나고 맨 아래에 있는 print(f'결과: {result}')가 실행됩니다.

이때 이번 실행에서는 result가 새로 만들어지지 못했기 때문에, 파이썬은 이전에 정상 실행되었을 때 메모리에 저장해 둔 예전 result 값을 끌어와서 보여준 것입니다.

````python
# try 문
try:
    file = open('data.csv', 'r')
    content = file.read( )
    print(content)

# except 문
except:
    print('파일을 찾을 수 없습니다.')

# finally 블럭
finally:
    print('파일 작업 종료.')
    file.close( )
````

finally는 에러가 나든 안 나든 무조건 실행되어야 하는 정리 작업(예: 열어둔 파일 닫기, 데이터베이스 연결 끊기 등)이 있을 때만 쓰는 문법입니다.

그냥 화면에 글자를 출력하는 print() 같은 작업이라면 finally를 굳이 쓰지 않고 try-except 바깥 맨 아래에 그냥 적어두어도 프로그램이 알아서 잘 처리합니다.

finally가 꼭 필요한 경우는 다음과 같은 상황들입니다.

open()으로 파일을 열어서 읽다가 도중에 에러가 터졌을 때, except로 빠지더라도 열린 파일을 반드시 닫아주어야(file.close()) 메모리 누수가 안 생깁니다.

네트워크 연결이나 데이터베이스 커넥션을 맺은 뒤 에러 여부와 상관없이 무조건 연결을 끊어주어야 할 때 사용합니다.

즉, 단순히 코드가 끝난 뒤 결과를 출력하거나 다음 단계를 실행하는 목적이라면 finally 없이 그냥 블록 바깥에 적어두셔도 아무 문제가 없습니다!

파일을 열 때는 예외가 났을 때 file.close()가 자동으로 실행되지 않기 때문에 finally나 별도의 조치가 없으면 파일이 닫히지 않습니다.

앞선 코드([23]번 셀)에서 예외가 났는데도 마지막 print(f'결과: {result}')가 실행되었던 이유는 예외가 발생해서 result라는 변수가 새로 만들어지지 못했을 뿐, 이전에 메모리에 살아있던 '옛날 결과값'을 그대로 가져와서 출력했기 때문입니다. 프로그램이 멈추지 않고 끝까지 내려왔던 것이죠.

하지만 파일 다루는 코드로 비교해 보면 상황이 완전히 다릅니다.

```
try:
    file = open('data.csv', 'r')  # 만약 여기서 에러가 나거나
    content = file.read()
    result = 10 / number          # 여기서 에러가 나면
    
except:
    print('에러 발생!')

# 만약 여기에 file.close()를 적어두었다면?
file.close()
```

만약 open()을 하다가 파일이 없어서 곧바로 에러가 터지거나, 파일을 여는 과정에서 문제가 생기면 file이라는 변수 자체가 아예 만들어지지 않거나 정상적인 파일 객체가 담기지 않습니다. 그런 상태로 맨 아래에 file.close()를 적어두면, 파이썬은 "만든 적도 없는 파일을 뭘 닫으라는 거냐"라며 NameError를 뿜어내며 프로그램이 또 터져버립니다.

그렇기 때문에:

정상적으로 파일을 열었을 때와 중간에 에러가 났을 때 어떤 상황이든 안전하게 파일을 닫으려면,

에러가 나든 안 나든 무조건 실행되는 보장 구역(finally)이 필요하고,

그 안에서 file 변수가 안전하게 만들어졌는지 확인하고 닫아주어야 안전하기 때문에 finally와 file.close()를 세트로 쓰는 것입니다.

요약하자면, 앞선 print문은 단순히 옛날 메모리 값을 출력해서 눈에 '실행된 것처럼' 보인 것뿐이고, 파일 자원은 예외가 나면 건너뛰어져 버리기 때문에 방치되거나 추가 에러를 유발하므로 finally로 확실하게 닫아주어야 하는 것입니다!
